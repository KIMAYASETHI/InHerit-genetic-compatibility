**RECURSIVE FUNCTIONS**

these are the functions that calls itself to solve a similar version of the same problem. basically a function that calls itself ...


need? eg- factorial of a number

5!=?

5!=5x4x3x2x1

5!=60

```python
#count down from 5 to 1 :-

#defining the function of a countdown
def countdown(n):
  print(n)

  if n>1:
    #calling the same"countdown" function
    countdown(n-1)
```

```python
#calling the function:

countdown(5)
```

every recursion or recursive functuion has two components:-


*   base case- when should we stop ?
*   recursive case- how do we make the problem smaller ?

recursion is like going down a staircase. the recursive call takes you to the next step, while the base makes you reach the ground.

```python
#using recursion find the factorial of a number :-

def factorial(n):
  if n==1:
    return 1
  else:
    return n * factorial(n-1)
```

```python
#calling the above written function:-

factorial(20)
```

```python
#error :-
age=int(input("enter your age "))
print("your age is ",age)
```

```python
#handling the error :-

try:

  age=int(input("enter your age"))
  print("your age is ",age )

except ValueError:
  print("please enter a valid age ")
```
