# Lists

**A list is simply a collection of items kept together in a particular order.**

````python
#creating a list
bacteria = ["E. coli", "Bacillus", "Salmonella", "Lactobacillus"]
````

````python
#printing a list
bacteria
````

````python
#accessing list items
print(bacteria[2])
````

````python
bacteria[1:3]
````

````python
#accessing list items
print(bacteria[2])
````

````python
#changing a list
bacteria = ["E. coli", "Bacillus", "Salmonella"]
bacteria
````

````python
#changing list item
bacteria[1] = "Staphylococcus"
print(bacteria)
````

````python
#adding list item.
#append means put one more item at the end
bacteria.append("Lactobacillus")
print(bacteria)
````

````python
#removing list items
bacteria.remove("Lactobacillus")
print(bacteria)
````

````python
#removing list items
bacteria.pop()
bacteria
````



```
remove() → remove a specific item
pop() → remove an item, normally from the end, but you can remove items using it's index as well
```



````python
#another example
numbers = [10, 20, 30, 40, 50]
numbers.pop(2)

print(numbers)
````

````python
#another way of removing an item using del
numbers = [10, 20, 30, 40, 50]

# Delete the element at index 2 (the number 30)
del numbers[2]

print(numbers)
````

````python
numbers = [10, 20, 30, 40, 50]

# Remove elements from index 1 up to (but not including) index 4
del numbers[1:4]

print(numbers)
````

Sometimes we have a collection of information that should NOT be changed.

A tuple uses parentheses.
coordinates = (10.5, 76.2)

List versus Tuples

**List**

A shopping list:

    You can add/remove/change things.

**Tuple**

A date of birth:

    It is a fixed piece of information.

````python
#defining a tuple
dob = (15, 8, 2001)
````

````python
#accessing a tuple
print(dob[0])
````

````python
#execute this and see the output
dob[0] = 20
````

````python
#another example
dimensions = (10, 20, 30)

print(dimensions)
print(dimensions[0])
print(dimensions[1])
````

**Lists are mutable. Tuples are immutable**

````python
#another example
sample = ("S001", "E. coli", 37)

#S001       → Sample ID
#E. coli    → Organism
#37         → Temperature
````

````python
print(sample[0])
print(sample[1])
print(sample[2])
````

````python
sample[0] = "demo"
````

Difference between List and Tuples

| Feature               | List                        | Tuple                        |
| --------------------- | --------------------------- | ---------------------------- |
| Brackets              | `[ ]`                       | `( )`                        |
| Ordered?              | Yes                         | Yes                          |
| Can change items?     | **Yes**                     | **No**                       |
| Can add/remove items? | **Yes**                     | **No**                       |
| Example               | `genes = ["BRCA1", "TP53"]` | `coordinates = (10.5, 76.2)` |
| Best analogy          | Shopping list               | Fixed record                 |

````python
#LIST

genes = ["BRCA1", "TP53", "MYC"]

genes[0] = "EGFR"

print(genes)
````

````python
#Tuple

coordinates = (10.5, 76.2)

coordinates[0] = 11.0
````



```
# List → ordered + changeable

# Tuple → ordered + unchangeable
```



# Dictionary

A dictionary stores information as a KEY → VALUE pair.

````python
#Think of it like a biological sample record.
#Instead of remembering that the organism is the second item, you give that information a meaningful name.

sample = {
    "sample_id": "S101",
    "organism": "E. coli",
    "temperature": 37,
    "pH": 7.0
}
````

````python
print(sample)
````

KEY             VALUE
-------------------------
**sample_id   →   S101**

**organism    →   E. coli**

**temperature →   37**

**pH          →   7.0**

````python
#With a list, we use a position:
organisms = ["E. coli", "Bacillus", "Salmonella"]
print(organisms[0])
````

````python
#With a dictionary, we use a key:
print(sample["temperature"])
````

**List → "Give me item number 1."**
**Dictionary → "Give me the value associated with 'organism'."**

````python
#managing entries in a dictionary


#We can change a value:
sample["pH"] = 7.2
````

````python
sample["gram_stain"] = "Gram-negative"
````

````python
sample
````

````python
#removing item
del sample["temperature"]
````

**Biological application: bacterial growth**

````python
#Suppose optical density (OD600) was measured during bacterial growth:
od = np.array([0.10, 0.18, 0.32, 0.55, 0.82, 1.10])
````

````python
#What was the average OD?
print(np.mean(od))
````

````python
#What was the maximum OD?
print(np.max(od))
````

````python
#How much did OD increase from the first to the last measurement?
increase = od[-1] - od[0]

print(increase)
````

**Array comparisons**

````python


print(od > 0.5)

#NumPy can check every measurement against a condition
````

````python
print(od[od > 0.5])
````

**Filtering with NumPy**

````python
import numpy as np
colony_counts = np.array([45, 120, 250, 80, 310, 175])

high_counts = colony_counts[colony_counts != 250]

print(high_counts)
````

````python
# & means both the conditions must be satisfied
#| means that either of the stated condistions should be satisfied
````
